```python
import os, sys#Import os for file-related operations and sys to exit the program

fname = input("enter file name=")#ask user to enter file name
if os.path.isfile(fname):#check given file exit
  f = open(fname, 'r')#open file in read mode
else:#if file doesn't exit
  print(fname+"does not exist")#display a message that file not exit
  sys.exit()#exit program

cl = cw = cc = 0#initialize countline,count wods,count characters
for line in f:
  words = line.split()
  cl+= 1
  cw+= len(words)
  cc+= len(line)

print("number of lines", cl)
print("number of words", cw)
print("number of characters", cc)

f.close()
```

```python
import zipfile
#step 1: Create a simple txt file to test with
with open("my_note.txt", "w") as f:
  f.write("Hello class 8! This is a test note.")
#step 2: open a new zip file "w" (write) node so we get zip file with below name
zip_bag = zipfile.ZipFile("note_backup.zip", "w")
#step 3: put our note file inside the zip bag which is variable
zip_bag.write("my_note.txt") #passing name of file which we want to zip
#step 4: Close the zip bag so it saves properly
zip_bag.close()
#step 5: Tell the user It worked!
print("Success! 'my_note.txt is nnow packed inside 'note_backup.zip'.")
```

```python
import zipfile
#Step 1: Open our zip file in 'r' (read) mode
zip_bag = zipfile.ZipFile("note_backup.zip", "r")
#Step 2: Extract (unzip) everything into a new floder named 'my_unzipped_note'
zip_bag.extractall("my_zippede_note") #so unzip files comes under this floder
#step 3: close the zip file
zip_bag.close()
#step 4: Tell the user it worked!
print("Success! File unzipped inside the 'my_unzipped' floder.")
```
